### lidar 2020

```python
#gs.run_command("r.mask", flags="r", quiet=True)
```

```python
gs.run_command(
    'v.in.pdal',
    input="lidar_2020.laz",
    output="dtm_2020_points",
    class_filter="2,11",  # Ground and Roads
    flags="w",
    overwrite=True
)
```

```python
!g.region vector="dtm_2020_points", res=1
```

```python
gs.run_command(
    "v.hull",
    input="dtm_2020_points",
    output="hull_2020",
    flags="f",
    quiet=True
)
```

```python
m = gj.Map(use_region=True)
m.d_vect(map="hull_2020")
m.show()
```

```python
gs.run_command(
    "v.to.rast",
    input="hull_2020",
    output="hull_2020_mask",
    use="value",
    type="area",
    value=1
)
```

```python
gs.run_command(
    "r.mask",
    raster="hull_2020_mask",
    overwrite=True
)
```

```python
gs.run_command(
    "v.surf.rst",
    input="dtm_2020_points",
    elevation="dtm_2020",
    smooth=1,
    dmin=2,
    segmax=40,
    tension=20,
    npmin=300,
    overwrite=True
)
```

```python
gs.run_command(
        "r.fillnulls",
        input="dtm_2020",
        output="dtm_2020_filled",
        method="rst",
        tension=10,
        smooth=1,
        edge=15,
        npmin=100,
        segmax=80,
        memory=5000,
        overwrite=True,
    )
```

```python
gs.run_command("r.relief", input="dtm_2020_filled", output="dtm_2020_filled_relief")
```

```python
m = gj.Map(use_region=True)
m.d_shade(color="dtm_2020_filled", shade="dtm_2020_filled_relief", brighten=50)
m.show()
```

### Export this DEMs into one folder

```python
# Create output folder
output_dir = "/home/pregmi3/CODEM/demo/All_elevation/codem_input"
os.makedirs(output_dir, exist_ok=True)
```

```python
export_list = {
    "dtm_2017": "lidar_2017_dtm.tif",
    "dtm_2020_filled": "lidar_2020_dtm.tif",
    "dtm_2024_filled": "lidar_2024_dtm.tif",
    "cap_dtm_2024_filled": "cap_dtm_2024.tif",
}

common_res = 1  # use 1 meter, or whatever resolution you want

for raster_name, output_name in export_list.items():

    # Set full extent of raster, but force square cells
    gs.run_command(
        "g.region",
        raster=raster_name,
        res=common_res,
        flags="a"
    )

    # Export using current region
    gs.run_command(
        "r.out.gdal",
        input=raster_name,
        output=os.path.join(output_dir, output_name),
        format="GTiff",
        createopt="COMPRESS=LZW",
        overwrite=True
    )

    print(f"Exported with square {common_res} m pixels: {output_name}")
```
